```python
dataset_path = "/content/plant_dataset/Plant_Disease_Dataset"

train_path = f"{dataset_path}/train"
valid_path = f"{dataset_path}/valid"
test_path  = f"{dataset_path}/test"
```

```python
import tensorflow as tf

print("TensorFlow:", tf.__version__)
print("GPU:", tf.config.list_physical_devices("GPU"))
```

```python
IMG_SIZE = (128, 128)
BATCH_SIZE = 32

training_set = tf.keras.utils.image_dataset_from_directory(
    train_path,
    labels="inferred",
    label_mode="categorical",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True
)

validation_set = tf.keras.utils.image_dataset_from_directory(
    valid_path,
    labels="inferred",
    label_mode="categorical",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)
```

```python
class_names = training_set.class_names

print("Number of classes:", len(class_names))
print(class_names)
```

```python
AUTOTUNE = tf.data.AUTOTUNE

training_set = training_set.prefetch(AUTOTUNE)
validation_set = validation_set.prefetch(AUTOTUNE)
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import (
    Input,
    Conv2D,
    MaxPool2D,
    Dropout,
    Flatten,
    Dense
)

model = Sequential([

    Input(shape=(128, 128, 3)),

    # Block 1
    Conv2D(32, 3, padding="same", activation="relu"),
    Conv2D(32, 3, activation="relu"),
    MaxPool2D(pool_size=2, strides=2),

    # Block 2
    Conv2D(64, 3, padding="same", activation="relu"),
    Conv2D(64, 3, activation="relu"),
    MaxPool2D(pool_size=2, strides=2),

    # Block 3
    Conv2D(128, 3, padding="same", activation="relu"),
    Conv2D(128, 3, activation="relu"),
    MaxPool2D(pool_size=2, strides=2),

    # Block 4
    Conv2D(256, 3, padding="same", activation="relu"),
    Conv2D(256, 3, activation="relu"),
    MaxPool2D(pool_size=2, strides=2),

    # Block 5
    Conv2D(512, 3, padding="same", activation="relu"),
    Conv2D(512, 3, activation="relu"),
    MaxPool2D(pool_size=2, strides=2),

    Dropout(0.25),

    Flatten(),

    Dense(1500, activation="relu"),
    Dropout(0.4),

    # 38 plant disease classes
    Dense(38, activation="softmax")
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
checkpoint = tf.keras.callbacks.ModelCheckpoint(
    "/content/best_model.keras",
    monitor="val_accuracy",
    save_best_only=True,
    verbose=1
)
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_accuracy",
    patience=3,
    restore_best_weights=True,
    verbose=1
)
```

```python
history = model.fit(
    training_set,
    validation_data=validation_set,
    epochs=10,
    callbacks=[
        checkpoint,
        early_stopping
    ]
)
```

```python
model.save("/content/trained_model.keras")
```

```python
import numpy as np
from tensorflow.keras.preprocessing import image

image_path = "/content/plant_dataset/Plant_Disease_Dataset/test/test/TomatoYellowCurlVirus2.JPG"

img = image.load_img(
    image_path,
    target_size=(128, 128)
)

img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)

prediction = test_model.predict(img_array)

predicted_index = np.argmax(prediction[0])
predicted_class = class_names[predicted_index]
confidence = prediction[0][predicted_index]

print("Prediction:", predicted_class)
print("Confidence:", f"{confidence:.2%}")
```
